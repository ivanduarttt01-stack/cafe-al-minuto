# ☕ Café al Minuto — Análisis exploratorio

```
import pandas as pd
import matplotlib.pyplot as plt
orders=pd.read_csv('../data/raw/orders.csv')
orders.head()
```

```
orders.describe(include='all').T
```

```
hourly=orders.groupby('hour').agg(pedidos=('order_id','count'),facturacion=('ticket_amount','sum'),espera=('wait_minutes','mean'))
hourly
```

```
hourly['facturacion'].plot(marker='o',figsize=(10,4)); plt.title('Facturación por hora'); plt.grid(alpha=.2); plt.show()
```

```
plt.figure(figsize=(9,5)); plt.scatter(orders.wait_minutes,orders.satisfaction,alpha=.18); plt.title('Espera vs satisfacción'); plt.xlabel('Espera'); plt.ylabel('Satisfacción'); plt.show()
```
